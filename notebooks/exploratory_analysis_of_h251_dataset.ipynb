# [MEPS HC 251 2023 Full Year Consolidated Data File ](https://meps.ahrq.gov/data_stats/download_data_files_detail.jsp?cboPufNumber=HC-251)
The Medical Expenditure Panel Survey (MEPS) Household Component (HS) is a nationally representative federal survey used by researchers and policymakers to measure healthcare in the U.S.. The survey was conducted by the U.S. Department of Health and Human Services's Agency for Healthcare Research and Quality (AHRQ) in 2023.

- What it measures: Healthcare use, medical expenses, payment sources, health insurance coverage, health status, and employment/demographics for the U.S. civilian population.

- How data is gathered: It tracks a large sample of households over time through computer interviews and cross-checks expenditure details directly with medical providers (doctors, hospitals, and pharmacies). 

- Dataset: The dataset contains microdata for 18,919 individuals interviewed throughout 2023.

| Role | MEPS variable | Meaning |
|---|---|---|
| Outcome | `TOTEXP23` | Total healthcare expenditures in 2023 |
| Main predictor | `PHYEXE53` | Whether the person does at least 30 minutes of moderate or vigorous physical activity five times a week |
| Main predictor | `OFTSMK53` | How often the person smokes cigarettes |
| Comparison variable | `AGE23X` | Age at the end of 2023 |
| Comparison variable | `SEX` | Sex |
| Comparison variable | `INSCOV23` | Health insurance coverage in 2023 |
| Optional comparison variable | `POVCAT23` | Family income relative to the poverty line |
| Survey variable | `PERWT23F` | Final 2023 person weight |
